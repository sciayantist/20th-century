# 03. Create a Dataset

```python
#Create a dataframe
animal_data = pd.DataFrame({"Animal Name": ['Monkey', 'Giraffe', 'Tiger', 'Penguin', 'Panda'],
                            "Favorite Foods": ['Bananas', 'Leaves', 'Deer', 'Fish', 'Bamboo'],
                            "Country": ['Egypt','Kenya', 'Russia', 'New Zealand', 'China'],
                            "Age": [ 2, 7, 3, 4, 5]})
                        
```

```python
animal_data.head(5)
```

# 04. Use Apply() & Lambda() Functions

```python
#Create a Triple Age function
def triple_age(row) : 
    return row['Age'] * 3
```

```python
#Apply the funtion to the column and save the values in a new column (deriving a variable)
animal_data['triple_Age'] = animal_data.apply(triple_age, axis =1)
```

```python
#Check the data
animal_data
```

```python
#Use Lambda Funstions animal_data
triple_Age_2 = lambda x:(x) * 3
```

```python
triple_Age_2
```

```python
#Check the column
triple_Age_2(animal_data['Age'])
```

```python
#Use the lambda funtion and save the values in a new column (deriving a variable)
animal_data['triple_Age_2'] = triple_Age_2(animal_data['Age'])
```

```python
#Check the data
animal_data
```

```python
#Create a list
Country_Region = {"Egypt":"North Africa",
                  "Kenya":"Africa",
                  "Russia":"Europe",
                  "New Zealand":"Oceania",
                  "China":"Asia"}
```

```python
#Check the list
Country_Region
```

```python
#Create a flag cloumn using lambda (deriving a variable)
animal_data['Region'] = animal_data['Country'].apply(lambda country: Country_Region.get(country))
```

```python
#Check the dataframe
animal_data
```

```python
#Alternative way
#Create a flag cloumn using lambda (deriving a variable)
animal_data['Region_2'] = animal_data['Country'].apply(
    lambda country: (
        'North Africa' if country == 'Egypt' else
        'Africa' if country == 'Kenya' else
        'Oceania' if country == 'New Zealand' else
        'Asia' if country == 'China' else
        'Unknown'))
```
